# __Su-Schrieffer-Heeger (SSH) Model II__

## _(1D Topological Insulator)_


The deep connection between **bulk topology** and the existence of **edge states** is examined—a defining feature of topological phases known as the **bulk–boundary correspondence**. In the Su–Schrieffer–Heeger (SSH) model, this correspondence is protected by a fundamental symmetry: **chiral symmetry**.

## Chiral Symmetry

A Hamiltonian $\hat{H}$ is said to possess a unitary symmetry $\hat{U}$ if  
$$
\hat{U} \hat{H} \hat{U}^\dagger = \hat{H}.
$$  
When such a symmetry is present, simultaneous diagonalization of $\hat{H}$ and $\hat{U}$ is possible, and the Hilbert space is partitioned into sectors labeled by the eigenvalues of $\hat{U}$. Dynamics governed by $\hat{H}$ are then confined within each sector.

In contrast, **chiral symmetry** is defined by the existence of a unitary operator $\hat{\Gamma}$ such that  
$$ \boxed{
\hat{\Gamma} \hat{H} \hat{\Gamma}^\dagger = -\hat{H}.}
$$  
This condition results in a spectrum that is symmetric about zero energy: for every eigenstate $|n\rangle$ with energy $E_n$, a corresponding eigenstate $\hat{\Gamma}|n\rangle$ with energy $-E_n$ is guaranteed.

The chiral symmetry operator $\hat{\Gamma}$ is required to satisfy three key conditions:

1. **Unitary and Hermitian**:  
   $$
   \hat{\Gamma}^\dagger = \hat{\Gamma}, \quad \hat{\Gamma}^2 = 1.
   $$  
   This ensures that $\hat{\Gamma}$ has eigenvalues $\pm 1$ and prevents the emergence of additional unitary symmetries from $\hat{\Gamma}^2$.

2. **Locality**:  
   The operator $\hat{\Gamma}$ is restricted to act within individual unit cells. For the SSH model, matrix elements between sites of different unit cells vanish:  
   $$
   \langle n, \alpha | \hat{\Gamma} | n', \alpha' \rangle = 0 \quad \text{for } n \ne n',
   $$  
   where $\alpha, \alpha' \in \{A, B\}$. The operator can be expressed as a tensor product over unit cells:  
   $$
   \hat{\Gamma} = \bigotimes_{n=1}^N \gamma,
   $$  
   with $\gamma$ acting on the internal sublattice space.

3. **Robustness**:  
   The symmetry must hold uniformly across all parameter realizations (e.g., including disorder). For a family of Hamiltonians $\{\hat{H}(\theta)\}$, it is required that  
   $$
   \hat{\Gamma} \hat{H}(\theta) \hat{\Gamma} = -\hat{H}(\theta) \quad \forall\, \theta,
   $$  
   with the same $\hat{\Gamma}$, independent of $\theta$.

### Sublattice Structure

The term **sublattice symmetry** is often used interchangeably with chiral symmetry. Projectors onto sublattices $A$ and $B$ are defined as  
$$
\hat{P}_A = \frac{1}{2}(1 + \hat{\Gamma}), \quad \hat{P}_B = \frac{1}{2}(1 - \hat{\Gamma}).
$$  
These satisfy $\hat{P}_A + \hat{P}_B = 1$ and $\hat{P}_A \hat{P}_B = 0$. Under chiral symmetry, the Hamiltonian is constrained such that  
$$
\hat{P}_A \hat{H} \hat{P}_A = \hat{P}_B \hat{H} \hat{P}_B = 0,
$$  
and thus only couples different sublattices:  
$$
\hat{H} = \hat{P}_A \hat{H} \hat{P}_B + \hat{P}_B \hat{H} \hat{P}_A.
$$  
This bipartite structure is inherent to the SSH model.

### Spectrum Symmetry

The energy spectrum of a chiral symmetric Hamiltonian is symmetric about zero. Given an eigenstate $|n\rangle$ with energy $E_n$, the state $\hat{\Gamma}|n\rangle$ is an eigenstate of $\hat{H}$ with energy $-E_n$, as shown by  
$$
\hat{H} \hat{\Gamma} |n\rangle = -\hat{\Gamma} \hat{H} |n\rangle = -E_n \hat{\Gamma} |n\rangle.
$$

- For $E_n \ne 0$: the states $|n\rangle$ and $\hat{\Gamma}|n\rangle$ are orthogonal and form a degenerate pair at $\pm E_n$. It follows that  
  $$
  \langle n | \hat{\Gamma} | n \rangle = \langle n | \hat{P}_A - \hat{P}_B | n \rangle = 0,
  $$  
  indicating that nonzero-energy eigenstates have equal weight on both sublattices.

- For $E_n = 0$: zero-energy eigenstates can be chosen as eigenstates of $\hat{\Gamma}$, and thus are supported entirely on a single sublattice. These states are invariant under the chiral symmetry operation up to a sign, and are referred to as self-conjugate.

### Chiral Symmetry in the SSH Model

The SSH Hamiltonian is inherently bipartite, containing only inter-sublattice hoppings. The sublattice projectors are given by  
$$
\hat{P}_A = \sum_{n=1}^N |n_A\rangle\langle n_A|, \quad \hat{P}_B = \sum_{n=1}^N |n_B\rangle\langle n_B|,
$$  
and the chiral symmetry operator is defined as  
$$
\hat{\Gamma} = \hat{P}_A - \hat{P}_B.
$$  
The anticommutation relation  
$$
\hat{\Gamma} \hat{H} \hat{\Gamma} = -\hat{H}
$$  
is satisfied, as each hopping term $|n_A\rangle\langle n'_B|$ acquires exactly one minus sign under conjugation by $\hat{\Gamma}$. This property is preserved even in the presence of spatially varying hopping amplitudes, ensuring the **robustness** of chiral symmetry.

## Bulk Winding Number

In momentum space, the bulk Hamiltonian is  
$$
\hat{H}(k) = \vec{d}(k) \cdot \vec{\sigma}, \quad \vec{d}(k) = (t1+ t2  \cos k, t2  \sin k, 0).
$$
* Chiral symmetry forces $d_z(k) = 0$, so $\vec{d}(k)$ lies in the $xy$-plane.

* As $k$ traverses the Brillouin zone, the tip of $\vec{d}(k)$ traces a closed loop.

* For an insulator, this loop must avoid the origin (otherwise the gap closes). 

* The number of times it winds around the origin is the **winding number** $\nu$, an integer topological invariant.

Thus, chiral symmetry enables a well-defined winding number in the bulk, which in turn determines the number of edge states at the boundary—establishing the **bulk–boundary correspondence**.

### Winding Number as an Integral

The winding number can also be written as a compact formula using the unit vector $\mathbf{d}_d$, defined as

$$
\mathbf{d}_d = \frac{\mathbf{d}}{|\mathbf{d}|}
$$

The vector $\mathbf{d}_d(k)$ is well defined for all $k$ because $\mathbf{d}(k) \ne 0$.

You can check easily that the winding number $\nu$ is given by

$$
\nu = \frac{1}{2\pi} \int_{-\pi}^{\pi} \left( \mathbf{d}_d(k) \times \frac{d}{dk} \mathbf{d}_d(k) \right)_z \, dk 
$$

To calculate $\nu$ directly from the bulk momentum-space Hamiltonian, note that it is off-diagonal (in the basis of eigenstates of the chiral symmetry operator $\sigma_z$),

$$
H(k) = \begin{pmatrix}
0 & h(k) \\
h^*(k) & 0
\end{pmatrix}, \quad \text{where } h(k) = d_x(k) - i d_y(k) 
$$

The winding number of $\mathbf{d}(k)$ can be written as an integral using the complex logarithm function, $\log(|h| e^{i \arg h}) = \log |h| + i \arg h$. It is easy to check that

$$
\nu = \frac{1}{2\pi i} \int_{-\pi}^{\pi} dk \, \frac{d}{dk} \log h(k)
$$

Here, during the calculation of the integral, the branch cut for the logarithm is always shifted so that the derivative is well defined. The above integral is always real, since $|h(k = -\pi)| = |h(k = \pi)|$.


```python
# Parameters
N = 20              # Number of unit cells
M = 2 * N           # Total number of sites
t1 = 0.5            # Intracell hopping (topological case: t1 < t2)
t2 = 1.0            # Intercell hopping
k_vals = np.linspace(-np.pi, np.pi, N)
```

```python
# bulk momentum-space Hamiltonian
def h(k, t1=1.0, t2=0.5):
    return t1 + t2 * np.exp(-1j * k)

# Winding number calc.
def compute_winding_number(t1=1.0, t2=0.5, N=1000):
    k_vals = np.linspace(-np.pi, np.pi, N, endpoint=False)
    hk = h(k_vals, t1, t2)

    # Compute log(h(k)) and its derivative
    log_hk = np.log(hk)
    dlog_dk = np.gradient(log_hk, k_vals)

    # Integrate
    winding_integral = np.trapz(dlog_dk, k_vals)
    winding_number = winding_integral / (2j * np.pi)

    return np.round(winding_number.real)
```

```python
# h(k) function
hk_vals = h(k_vals, t1, t2)

# winding number
winding = compute_winding_number(t1, t2)
print(f"winding number = {winding}")
```

```python
plt.figure(figsize=(6,6))
plt.plot(hk_vals.real, hk_vals.imag)
plt.plot(0, 0, 'ro', label='Origin')
plt.xlabel('Re[h(k)]')
plt.ylabel('Im[h(k)]')
plt.title('Trajectory of $h(k)$ in Complex Plane')
plt.grid(True)
plt.axis('equal')
plt.legend()
plt.show()
```

## Exact Calculation of Edge States

We now compute the zero-energy edge states of the SSH model in real space, without requiring translational invariance. Consider an SSH chain with $ N $ unit cells and complex intracell ($ t_1 $) and intercell ($ t_2 $) hopping amplitudes. The real-space Hamiltonian is:

$$
\hat{H} = \sum_{n=1}^{N} \left( t_{1,n} |n_B\rangle \langle n_A| + \text{h.c.} \right) + \sum_{n=1}^{N-1} \left( t_{2,n} |(n+1)_A\rangle \langle n_B| + \text{h.c.} \right)
$$

We look for a zero-energy eigenstate of the form:

$$
\hat{H} \left( \sum_{n=1}^{N} a_n |n_A\rangle + b_n |n_B\rangle \right) = 0
$$

This yields the following set of equations:

For $ n = 1, \dots, N-1 $:

$$
t_{1,n} a_n + t_{2,n} a_{n+1} = 0, \quad t_{2,n} b_n + t_{1,n+1} b_{n+1} = 0
$$

At the boundaries:

$$
t_{1,N} a_N = 0, \quad t_{1,1} b_1 = 0
$$

The recursive solution to the bulk equations is:

$$
a_n = a_1 \prod_{j=1}^{n-1} \left( -\frac{t_{1,j}}{t_{2,j}} \right), \quad b_n = b_N \cdot \left( -\frac{t_{1,N}}{t_{2,n}} \prod_{j=n+1}^{N-1} \frac{t_{1,j}}{t_{2,j}} \right)
$$

The boundary conditions enforce $ b_1 = a_N = 0 $, which implies that in the generic finite case, there is no exact zero-energy eigenstate unless some hoppings vanish.

However, approximate edge-localized solutions exist in the thermodynamic limit $ N \to \infty $, provided that the average intercell hopping dominates the intracell hopping. Define:

$$
\log |\bar{t}_1| = \frac{1}{N-1} \sum_{n=1}^{N-1} \log |t_{1,n}|, \quad \log |\bar{t}_2| = \frac{1}{N-1} \sum_{n=1}^{N-1} \log |t_{2,n}|
$$

Then the amplitudes decay exponentially as:

$$
|a_N| = |a_1| e^{-(N-1)/\xi}, \quad |b_1| = |b_N| e^{-(N-1)/\xi} \cdot \left| \frac{t_{1,N}}{t_{1,1}} \right|
$$

with localization length:

$$
\xi = \frac{1}{\log |\bar{t}_2| - \log |\bar{t}_1|}
$$

This length scale determines how quickly the wavefunction decays away from the edge. It depends on the **difference** between the average strengths of inter-cell ($ t_{2,n} $) and intra-cell ($ t_{1,n} $) hoppings.

- If $ \log |t_{2,n}| > \log |t_{1,n}| $ (inter-cell hoppings dominate), then $ \xi > 0 $, and edge states can exist.
- If $ \log |t_{2,n}| \leq \log |t_{1,n}| $ (intra-cell hoppings dominate or equal inter-cell hoppings), then $ \xi \leq 0 $, and no localized edge states form.

This shows that exponentially localized edge modes appear when $ |\bar{t}_2| > |\bar{t}_1| $, consistent with the topological phase of the SSH model.


```python
# Compute a_n recursively
a = np.zeros(N)
a[0] = 1.0  # Set a1 = 1

for n in range(1, N):
    a[n] = a[n-1] * (-t1 / t2)

# Compute b_n recursively (starts from b_N = 1)
b = np.zeros(N)
b[-1] = 1.0  # Set b_N = 1

for n in range(N-2, -1, -1):
    b[n] = b[n+1] * (-t1 / t2)

# Normalize for better plotting
a_norm = a / np.max(np.abs(a))
b_norm = b / np.max(np.abs(b))

# Create positions for plotting
positions = np.arange(1, N+1)
```

```python
# Plot the edge state amplitudes
plt.figure(figsize=(8, 5))
plt.plot(positions, a_norm, 's-', label=r'$a_n$', color='blue')
plt.plot(positions, b_norm, 's-', label=r'$b_n$', color='red')
plt.axhline(0, color='gray', linestyle=':', linewidth=0.7)
plt.xlabel(r'Site index $n$')
plt.ylabel(r'$|\psi|^2$')
plt.title(f'Edge State Amplitudes (N={N}, $t_1$={t1}, $t_2$={t2})')
plt.grid(True, linestyle=':')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Define site-dependent hoppings
t1_n = np.linspace(2.5, 0.1, N - 1)  # intracell hopping
t2_n = np.linspace(0.1, 2.5, N - 1)  # intercell hopping

# Compute logarithmic averages
log_t1_bar = np.mean(np.log(np.abs(t1_n)))
log_t2_bar = np.mean(np.log(np.abs(t2_n)))

# Localization length
xi = 1 / (log_t2_bar - log_t1_bar)

# Edge amplitudes (example: assume |a_1| = 1, |b_N| = 1)
a1 = 1.0
aN = a1 * np.exp(-(N - 1) / xi)

bN = 1.0
b1 = bN * np.exp(-(N - 1) / xi) * np.abs(t1_n[-1] / t1_n[0])
```

```python
# Display results
print(f"log(|t1_bar|) = {log_t1_bar:.2e}")
print(f"log(|t2_bar|) = {log_t2_bar:.2e}")
print(f"Localization length ξ = {xi:.2e}")
print(f"Edge amplitude a_N = {aN:.2f}")
print(f"Edge amplitude b_1 = {b1:.2f}")
```

### Hybridization of Edge States

The left and right edge-localized states can be written as:

$$
|L\rangle = \sum_{n=1}^N a_n |n_A\rangle, \quad |R\rangle = \sum_{n=1}^N b_n |n_B\rangle
$$

with the coefficients \$a\_n\$ and \$b\_n\$ chosen according to the recursion relations, and normalized by setting \$a\_1\$ and \$b\_N\$ appropriately.

In a finite system, these two edge states are not completely decoupled; they hybridize under the full Hamiltonian \$\hat{H}\$, leading to a **small energy splitting**. This effect is exponentially suppressed in the system size \$N\$.

To a good approximation, the energy splitting can be estimated by computing the overlap:

$$
\langle R | \hat{H} | L \rangle = \left| a_1 e^{-(N-1)/\xi} b_N \right| e^{i\phi}
$$

with some arbitrary phase \$\phi \in $$0, 2\pi)\$. The hybridized eigenstates are:

$$
|0^+\rangle = \frac{e^{-i\phi/2} |L\rangle + e^{i\phi/2} |R\rangle}{\sqrt{2}}, \quad E_+ = \left| a_1 e^{-(N-1)/\xi} b_N \right|
$$

$$
|0^-\rangle = \frac{e^{-i\phi/2} |L\rangle - e^{i\phi/2} |R\rangle}{\sqrt{2}}, \quad E_- = -\left| a_1 e^{-(N-1)/\xi} b_N \right|
$$

Thus, the **energy splitting** between these two hybridized states is:

$$
\Delta E = 2 \left| a_1 e^{-(N-1)/\xi} b_N \right|
$$

This energy gap is **exponentially small** in \$N\$, and provides a signature of the **edge-state interaction**. In the topological phase with large \$N\$, this hybridization vanishes and the edge states become **nearly exact zero modes**.

```python
# Overlap and energy splitting
overlap = np.abs(a1 * np.exp(-(N - 1) / xi) * bN)
delta_E = 2 * overlap
```

```python
print(f"Overlap |<R|H|L>| = {overlap:.2f}")
print(f"Energy splitting ΔE = {delta_E:.2F}")
```

## Bound States at Domain Walls

Edge states in the SSH model are not limited to the physical ends of an open chain; they can also appear at **domain walls** between topologically distinct insulating regions within the same chain.

This can be intuitively understood in the **fully dimerized limit**. Consider a system with alternating domains of different dimerization patterns. At the boundaries between these domains—called **domain walls**—new localized states can emerge.

There are two types of domain walls:

* A **single isolated site**, which hosts a **zero-energy state**:

  $$
  \hat{H} \left( |6_B\rangle - |7_B\rangle \right) = 0
  $$

* A **trimer**, where a zero-energy eigenstate is formed by an **odd superposition** of the two end sites.

Just like the edge states at the ends of the chain, these **domain-wall states** are **sublattice-polarized**—the wavefunction has support only on one sublattice.
![5](5.png) 

From a perfect dimerized configuration, **only an even number of domain walls can be introduced**. This means that if a domain wall localizes a state on one sublattice, there must be another domain wall somewhere else (possibly at the system edge) localizing a state on the opposite sublattice.

Now, consider a domain wall configuration **away from the fully dimerized limit**. The domain wall modes remain **exponentially localized**, but their wavefunctions now **penetrate slightly into the bulk**, forming **evanescent tails**. If two domain walls are separated by $M$ unit cells, their corresponding localized edge states **hybridize**, forming bonding and anti-bonding combinations. At **half-filling**, only the **lower-energy (bonding) state** is occupied.

This bonding state:

* Contains **one electron**,
* Has a wavefunction **symmetrically localized** on both domain walls,
* And therefore, **each domain wall carries half an electronic charge**.

This effect is a hallmark of **charge fractionalization**, and provides a striking consequence of the topological nature of the SSH model.


```python
# Define two regions (topological and trivial)
t1_left = 0.5
t2_left = 1.0
t1_right = 1.0
t2_right = 0.5

# Initialize Hamiltonian
H = np.zeros((M, M))

# Fill Hamiltonian with left region (topological)
for i in range(0, M//2 - 1, 2):
    H[i, i+1] = t1_left
    H[i+1, i] = t1_left

for i in range(1, M//2 - 1, 2):
    H[i, i+1] = t2_left
    H[i+1, i] = t2_left

# Fill Hamiltonian with right region (trivial)
for i in range(M//2, M - 1, 2):
    H[i, i+1] = t1_right
    H[i+1, i] = t1_right

for i in range(M//2 + 1, M - 1, 2):
    H[i, i+1] = t2_right
    H[i+1, i] = t2_right

# Diagonalize
eigenvalues, eigenvectors = np.linalg.eigh(H)

# Sort eigenvalues and reorder eigenvectors accordingly
sorted_indices = np.argsort(eigenvalues)
eigenvalues = eigenvalues[sorted_indices]
eigenvectors = eigenvectors[:, sorted_indices]
```

```python
# Plot eigenvalues
plt.figure(figsize=(6, 4))
plt.scatter(range(M), eigenvalues, color='darkblue', s=30)
plt.axhline(0, color='gray', linestyle='--', linewidth=0.8)
plt.xlabel('Eigenvalue Index')
plt.ylabel('Energy')
plt.title('SSH Model with Domain Wall')
plt.grid(True)
plt.show()
```

```python
# Find index of eigenvalue closest to 0
idx_zero = np.argmin(np.abs(eigenvalues))
bound_state = eigenvectors[:, idx_zero]

# Plot amplitude
plt.figure(figsize=(6, 4))
plt.plot(np.abs(bound_state)**2, 'o-', color='darkblue')
plt.xlabel('Site')
plt.ylabel(r'$|\psi|^2$')
plt.title('Domain Wall State Localization')
plt.grid(True)
plt.show()
```

```python
# Loop 1: Vary t1_left
eigs = []
for t1_left in t_vals:
    H = np.zeros((M, M))
    
    # Left region (first half)
    for i in range(0, M//2 - 1, 2):  # A_n -> B_n
        H[i, i+1] = t1_left
        H[i+1, i] = t1_left
    for i in range(1, M//2 - 1, 2):  # B_n -> A_{n+1}
        H[i, i+1] = t2_left_base
        H[i+1, i] = t2_left_base

    # Right region (second half)
    for i in range(M//2, M - 1, 2):  # A_n -> B_n
        H[i, i+1] = t1_right_base
        H[i+1, i] = t1_right_base
    for i in range(M//2 + 1, M - 1, 2):  # B_n -> A_{n+1}
        H[i, i+1] = t2_right_base
        H[i+1, i] = t2_right_base

    eigvals = np.linalg.eigvalsh(H)
    eigs.append(eigvals)

eigs_list.append(np.array(eigs).T)

# Loop 2: Vary t2_left
eigs = []
for t2_left in t_vals:
    H = np.zeros((M, M))
    
    # Left region
    for i in range(0, M//2 - 1, 2):
        H[i, i+1] = t1_left_base
        H[i+1, i] = t1_left_base
    for i in range(1, M//2 - 1, 2):
        H[i, i+1] = t2_left
        H[i+1, i] = t2_left

    # Right region
    for i in range(M//2, M - 1, 2):
        H[i, i+1] = t1_right_base
        H[i+1, i] = t1_right_base
    for i in range(M//2 + 1, M - 1, 2):
        H[i, i+1] = t2_right_base
        H[i+1, i] = t2_right_base

    eigvals = np.linalg.eigvalsh(H)
    eigs.append(eigvals)

eigs_list.append(np.array(eigs).T)

# Loop 3: Vary t1_right
eigs = []
for t1_right in t_vals:
    H = np.zeros((M, M))
    
    # Left region
    for i in range(0, M//2 - 1, 2):
        H[i, i+1] = t1_left_base
        H[i+1, i] = t1_left_base
    for i in range(1, M//2 - 1, 2):
        H[i, i+1] = t2_left_base
        H[i+1, i] = t2_left_base

    # Right region
    for i in range(M//2, M - 1, 2):
        H[i, i+1] = t1_right
        H[i+1, i] = t1_right
    for i in range(M//2 + 1, M - 1, 2):
        H[i, i+1] = t2_right_base
        H[i+1, i] = t2_right_base

    eigvals = np.linalg.eigvalsh(H)
    eigs.append(eigvals)

eigs_list.append(np.array(eigs).T)

# Loop 4: Vary t2_right
eigs = []
for t2_right in t_vals:
    H = np.zeros((M, M))
    
    # Left region
    for i in range(0, M//2 - 1, 2):
        H[i, i+1] = t1_left_base
        H[i+1, i] = t1_left_base
    for i in range(1, M//2 - 1, 2):
        H[i, i+1] = t2_left_base
        H[i+1, i] = t2_left_base

    # Right region
    for i in range(M//2, M - 1, 2):
        H[i, i+1] = t1_right_base
        H[i+1, i] = t1_right_base
    for i in range(M//2 + 1, M - 1, 2):
        H[i, i+1] = t2_right
        H[i+1, i] = t2_right

    eigvals = np.linalg.eigvalsh(H)
    eigs.append(eigvals)

eigs_list.append(np.array(eigs).T)
```
